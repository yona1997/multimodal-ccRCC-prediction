```python
import os
import numpy as np

features_dir = '/media/Data/Yona/WSI CCRCC/features'
output_pool_dir = '/media/Data/Yona/WSI CCRCC/features_MeanPooling'
os.makedirs(output_pool_dir, exist_ok=True)
npz_files = [os.path.join(features_dir, f) for f in os.listdir(features_dir) if f.endswith('.npz')]

for npz_file in npz_files:
    base_name = os.path.splitext(os.path.basename(npz_file))[0]
    out_filename = os.path.join(output_pool_dir, f"{base_name}-meanP.npz")
    if os.path.exists(out_filename):
        print(f"{os.path.basename(npz_file)} already processed, skipping.")
        continue
    data = np.load(npz_file)
    if 'features' not in data:
        print(f"Warning: 'features' key missing in {npz_file}.")
        continue
    features = data['features']
    if features.size == 0:
        print(f"Warning: empty features array in {npz_file}, skipping.")
        continue
    pooled_vector = np.mean(features, axis=0)
    np.savez_compressed(out_filename, features=pooled_vector)
    print(f"{os.path.basename(npz_file)} -> {os.path.basename(out_filename)}")
```

```python
import os
import numpy as np

features_dir = '/media/Data/Yona/WSI CCRCC/features'
output_pool_dir = '/media/Data/Yona/WSI CCRCC/features_MaxPooling'
os.makedirs(output_pool_dir, exist_ok=True)

npz_files = [os.path.join(features_dir, f) for f in os.listdir(features_dir) if f.endswith('.npz')]

for npz_file in npz_files:
    base_name = os.path.splitext(os.path.basename(npz_file))[0]
    out_filename = os.path.join(output_pool_dir, f"{base_name}-maxP.npz")
    
    if os.path.exists(out_filename):
        print(f"{os.path.basename(npz_file)} has already been processed")
        continue

    data = np.load(npz_file)
    
    if 'features' not in data:
        print(f"Warning: 'features' key is missing in {npz_file}.")
        continue

    features = data['features']
    
    if features.size == 0:
        print(f"Warning: empty 'features' table in {npz_file}, file ignored.")
        continue
    
    pooled_vector = np.max(features, axis=0)
    
    np.savez_compressed(out_filename, features=pooled_vector)
    print(f"{os.path.basename(npz_file)} -> {os.path.basename(out_filename)}")
```
